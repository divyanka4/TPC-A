```sql
CREATE CATALOG IF NOT EXISTS tpca
  COMMENT 'TPC-A project catalog (DEV)';
```

```sql
CREATE SCHEMA IF NOT EXISTS tpca.bronze
    COMMENT "bronze zone for raw tables with metadata";
CREATE SCHEMA IF NOT EXISTS tpca.silver
    COMMENT "silver zone for transformed data";
CREATE SCHEMA IF NOT EXISTS tpca.gold
    COMMENT "gold zone for final tables";
```

```sql
CREATE VOLUME IF NOT EXISTS tpca.bronze.raw;
CREATE VOLUME IF NOT EXISTS tpca.bronze.pipeline_state;
```
